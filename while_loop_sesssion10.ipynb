- any loop we required three things
  
- intial point (start point)
  
- increment or decrement
  
- condition to stop

- in for loop the above three happens in a single line

- but in while loop all 3 needs 3 different lines

```python
# start point 
# while <condition>:
#       <start our code >
#       increment and decrement 


range(1,10,1)

i=1
while <condition>:
    i=i+1
```

```python
i=1
while i<10:
    print(i)
    i=i+1
```
