**if-elif-else**

- when the is a condition is there
- we can expect two answers
    - in that first answer under if block
    - second answer under else block
- when there are two conditions
- we csn expect three answers
   - in that first answer under **if** block based on condition 1
   - second answer under **elif**block based on condition 2
   - third answer under **else** block     

```python
one conditions ==== two answer==== if else 
two conditions === 3 answers=====if-elif-else
if elif need conditions 
```

```python
if<cond>:
    <write your code>
elif <cond>:
    <write your code>
else:
    <write your code>
```

```python
# wap a program ask user enter a num
# if num equal to 1 print one 
#otherwise if number qual to 2 print two
# otherwise if number qual to 3 then print three 
#otherwise print enter a valid number 


num=eval(input('enter a number'))
if num==1:
    print('one')
elif num==2:
     print('two')
elif num==3:
    print('three')
else:
    print('enter a valid number')
```

```python
#wap ask user enter percentage 
# if percentage greater than 75 print A grade 
# elif precentage 60 to 75 print b grade 
# elif print p between 40 to 60 print c grade 
# elif p btween 30to 40 print d grade 
# otherwise print fail 



per=eval(input('enter the percentage:'))
if per>=75:
    print('A Grade')
elif per>=60:
    print('B Grade')
elif per >=45:
    print('C Grade')
elif per>=30:
    print('D grade')
else:
    print('fail')
```
