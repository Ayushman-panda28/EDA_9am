**case-3:range(start,stop,step)**

- start value = start
- if step is postive sign 
  - it is positive direction
  - increment by **step value**
  - last value = stop-1

- if step is negative sign
   - it is negative direction(decrement)
   - decrement by **step value**
   - last value=stop+1

```python
 for i in range(1,20,2):
     print(i,end=' ')
# start=1
# step=+2
# increment by 2 
# last stop-1 = 20-2=18
```

**continue-pass-break**

```python
for i in range(5):
    if i==2:
        print('great')
        pass
    print(i,end=' ')
```

```python
for i in range(5):
    if i==2:
        print('great')
        continue
    print(i,end=' ')
```

```python
#wap a program ask the user get 5 random numbers 
# apply even or odd logic 
# and count how many even numbers are appered 


import random
even_count=0
odd_count=0
even_summ=0
odd_summ=0
for i in range(5):
        num=random.randint(1,100)
        if num%2==0:
            even_count=even_count+1
            even_summ=even_summ+num
            print(f' {num} is even')
        else:
            odd_count=odd_count+1
            odd_summ=odd_summ+num
            print(f'{num} is odd')

print(f'the total even count is:{even_count}')
print(f'the total odd count is:{odd_count}')
print(even_summ,odd_summ)
```
