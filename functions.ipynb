**args-kwargs**

- args means we can provide multiple values without arguments name
 - which means calling function we can provide directly 'virat',102
 - while define the function you no need to mention arguments names  also
 - the out will come in **tuples** format:('virat',102)
- kwargs keyword arguments means we can provide multiple values with arguments name
   - while defines the function no need to provide the name of the arguments
   - while call the functions definitely we need to provide with arguments names
   - name='virat',runs=102
   - the output in dictornary format: {'name':'virat','runs':102}

```python
#def fun1(a,b):
   # print(a)
   # print(b)
#fun(10,20)


def fun1(*args):
    print(args)
fun1(10,20)
#without name 
# out in tuple format 
```

```python
def fun2(**kwargs):
    print(kwargs)
fun2(a=10,b=20)
#with name 
# output in dictornary format
```

```python
def f1(n):
    print(n)
    if n==0:
        return(0)
    f1(n-1)
f1(10)
```

```python
def f2(n):
    if n==0:
        return(0)
    add=n+f2(n-1)
    return(add)
f2(5)
```

```python
def fact(n):
    if n==0:
        return(1)
    fact=n*fact(n-1)
    return(fact)
fact(5)
```
